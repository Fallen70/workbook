# Classes 

Python est un language de Programation Orienté Objet, même s'il est trés efficace pour faire du scripting. Il posséde également de nombreux outils pour créer des `Class`.  
Une `Class` permet de créer des objets, un objet permet de regrouper des données (un état) et des méthodes (fonctions pour exploiter / modifier les l'état de l'objet).  

## Creation d'une classe

Pour créer une `Class` en python il faut utiliser le mot clef `class` suivi du nom de la classe puis `:`. Le nom de la classe est habituellement en CamelCase `Commence par une lettre majuscule pas de tiret ou sous-tiret et chaque mot commence par une lettre majuscule`.
example:
 * CamelCase
 * SnowLeopard

````python
class Product:

    def __init__(self, name, cost, price):
        self.name = name
        self.cost = cost
        self.price = price

    def profit_margin(self):
        return self.price - self.cost

duck = Product(name="rubber duck", cost=1, price=5)
print("Duck => ", duck )
print( "Type => ", type(duck) )
print( "Profit =>", duck.profit_margin() )
````

## Self

Le paramétre `self` dans les fonctions ci-dessus et un pointeur vers l'instance de la classe ( ici duck ). On peut voir ci-dessous qu'ils partagent la même référence.  
`self` n'est pas un mot réservé par python 

````python
class TestId:

    def ref(self):
        return id(self)

my_id = TestId()
print( "id(my_id) \t:", id(my_id) )
print( "my_id.ref() \t:", my_id.ref() )
````

## decorateur et dunder method

### dunder methods

Il y a de multitude `dunder methods` leur nom est entourée de `__` ( ex: `__str__`, `__eq__`, `__repr__`, `__lt__`, etc...)
Elles permettent de donner un comportement a l'objet que ce soit sa représentation lors d'un print ou l'égalité/infériorité avec un autre objet


````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    def __eq__(self, other):
        return (self.x,self.y) == (other.x,other.y)

p1 = Point(1,2)
p2 = Point(0,0)
if p1 != p2:
    print( f"{p1} and {p2} are not the same" )
````

### Décorateurs

Voici quelques `decorator` utiles :

* **property** : Permet d'accéder à une méthode comme si c'était un attribut.
* **attribut.setter** : Permet de définir un setter pour protéger l'écriture d'un attribut ou pour modifier un attribut qui n'est pas directement accessible.
* **classmethod** : Permet de définir une méthode de classe, qui est associée à la classe elle-même plutôt qu'à une instance spécifique.

Par exemple, nous pouvons réécrire la classe Product pour en faire une fabrique qui conserve un historique des instances créées, tout en utilisant certains des décorateurs mentionnés ci-dessus.

````python
class Product:

    products = []

    def __init__( self, name, cost, price ):
        self.name = name
        self.cost = cost
        self.price = price

    def __new__(cls, name, cost, price):
        self = super().__new__( cls )
        cls.products.append(self)
        return self
    
    def __repr__(self):
        return f"<Product:{self.name}, price: {self.price}>"
    
    @property
    def profit_margin(self):
        return self.price - self.cost

    @classmethod
    @property
    def total_products(cls):
        return len(cls.products)
        
print( Product.products )
duck = Product(name="rubber duck", cost=1, price=5)
print( Product.products )
pencil = Product(name="pencil", cost=1, price=3)
print( Product.products )
````
